```python
import pandas as pd
df = pd.read_csv("googleplaystore_user_reviews.csv")
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df.isna().sum()
```

```python
df_clean=df.copy()
```

```python
#drop na value
df_clean = df.dropna(subset=['Translated_Review']).copy()
```

```python
#remove duplicate
df_clean = df_clean.drop_duplicates()
df_clean.duplicated().sum()
```

```python
df_clean.isna().sum()
```

```python
#remove spases
df_clean['App'] = df_clean['App'].str.strip()
df_clean['Translated_Review'] = df_clean['Translated_Review'].str.strip()
```

```python
#check values between 1:-1
df_clean = df_clean[
    (df_clean['Sentiment_Polarity'] >= -1) & (df_clean['Sentiment_Polarity'] <= 1) &
    (df_clean['Sentiment_Subjectivity'] >= 0) & (df_clean['Sentiment_Subjectivity'] <= 1)
]
```

```python
#arrange row number
df_clean.to_csv('cleaned_googleplaystore_user_reviews.csv', index=False)
```
